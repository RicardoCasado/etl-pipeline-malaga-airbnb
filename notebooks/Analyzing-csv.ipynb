B. Comprobar la unicidad de las claves primarias

Este código responde directamente a una de las preguntas más importantes del proyecto: ¿podemos definir estas columnas como PRIMARY KEY en PostgreSQL?

Si encontramos duplicados, no los borraremos inmediatamente. Primero investigaremos por qué existen.

```python

df_listings = pd.read_csv(
    "../dataset/listings.csv", low_memory=False
)

df_reviews = pd.read_csv(
    "../dataset/reviews.csv", low_memory=False
)

df_calendar = pd.read_csv(
    "../dataset/calendar.csv", low_memory=False
)

def check_key(df, columns):
    print("Key:", columns)
    print("Null keys:", df[columns].isna().any(axis=1).sum())
    print("Duplicate keys:", df.duplicated(subset=columns).sum())
    print("Unique:", not df[columns].isna().any().any()
          and not df.duplicated(subset=columns).any())

check_key(df_listings, ["id"])
check_key(df_reviews, ["id"])
check_key(df_calendar, ["listing_id", "date"])
```

C. Comprobar la integridad referencial

Este análisis comprueba si hay reseñas y registros de calendario que apuntan a anuncios inexistentes en listings.

Si existen, tendremos que tomar una decisión de ingeniería: conservarlos en la capa raw, documentar el problema y determinar qué hacer con ellos en nuestros modelos transformados.

Es importante porque una clave foránea de PostgreSQL podría impedir insertar esos registros si no existe el correspondiente alojamiento.

```python

listing_ids = set(df_listings["id"].dropna())

reviews_orphans = (
    ~df_reviews["listing_id"].isin(listing_ids)
).sum()

calendar_orphans = (
    ~df_calendar["listing_id"].isin(listing_ids)
).sum()

print("Reviews without listing:", reviews_orphans)
print("Calendar rows without listing:", calendar_orphans)
```

```python

listing_ids = set(df_listings["id"])
review_listing_ids = set(df_reviews["listing_id"])

common_ids = listing_ids & review_listing_ids
orphan_ids = review_listing_ids - listing_ids

print("Listings in current snapshot:", len(listing_ids))
print("Listings represented in reviews:", len(review_listing_ids))
print("Matching listing IDs:", len(common_ids))
print("Unmatched listing IDs:", len(orphan_ids))

# Filas de reviews
matched_reviews = df_reviews[
    df_reviews["listing_id"].isin(common_ids)
]

orphan_reviews = df_reviews[
    df_reviews["listing_id"].isin(orphan_ids)
]

print("Matched review rows:", len(matched_reviews))
print("Unmatched review rows:", len(orphan_reviews))
print(
    "Percentage unmatched:",
    round(len(orphan_reviews) / len(df_reviews) * 100, 2)
)
```

3. Siguiente paso: estudiar las 90 columnas de listings.csv

```python

import pandas as pd

df_listings = pd.read_csv(
    "../dataset/listings.csv",
    low_memory=False
)

column_report = pd.DataFrame({
    "column": df_listings.columns,
    "dtype": df_listings.dtypes.astype(str).values,
    "missing_pct": (
        df_listings.isna().mean().values * 100
    ).round(2),
    "unique_values": df_listings.nunique().values
})

pd.set_option("display.max_rows", 100)

display(column_report)
```

Es decir, necesitamos comprobar dependencias funcionales como:
\[
\text{host\_id}\rightarrow\text{host\_is\_superhost}
\]

```python

host_columns = [
    "host_name",
    "host_profile_id",
    "host_location",
    "host_is_superhost",
    "host_listings_count",
    "host_identity_verified"
]

host_consistency = {}

for col in host_columns:
    inconsistent_hosts = (
        df_listings.groupby("host_id")[col]
        .nunique(dropna=False)
        .gt(1)
        .sum()
    )

    host_consistency[col] = inconsistent_hosts

display(
    pd.DataFrame.from_dict(
        host_consistency,
        orient="index",
        columns=["inconsistent_hosts"]
    )
)
```

4. Columnas completamente vacías

```python

empty_columns = column_report.loc[
    column_report["missing_pct"] == 100,
    "column"
].tolist()

print("Completely empty columns:", len(empty_columns))

for col in empty_columns:
    print(col)
```

```python

remaining_host_columns = [
    "host_url",
    "host_about",
    "host_picture_url",
    "hosts_time_as_user_years",
    "hosts_time_as_user_months",
    "hosts_time_as_host_years",
    "hosts_time_as_host_months",
    "host_has_profile_pic"
]

results = []

for col in remaining_host_columns:
    inconsistent = (
        df_listings.groupby("host_id")[col]
        .nunique(dropna=False)
        .gt(1)
        .sum()
    )

    results.append({
        "column": col,
        "inconsistent_hosts": inconsistent,
        "missing_pct": round(
            df_listings[col].isna().mean() * 100, 2
        )
    })

host_report = pd.DataFrame(results)

display(host_report)
```

4. Segunda comprobación: convertir el precio para analizarlo

```python

# Conversion for exploratory analysis only

price_numeric = pd.to_numeric(
    df_listings["price"]
    .astype("string")
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

display(price_numeric.describe())

print("Missing prices:", price_numeric.isna().sum())

print("Prices equal to zero:", (price_numeric == 0).sum())

print("Negative prices:", (price_numeric < 0).sum())

print(
    "New missing values after conversion:",
    (price_numeric.isna() & df_listings["price"].notna()).sum()
)
```

5. Tercera comprobación: comparar los dos precios

```python

price_comparison = pd.DataFrame({
    "price": price_numeric,
    "quoted_price_per_night":
        df_listings["price_quote_price_per_night"],
    "quoted_total":
        df_listings["price_quote_total_price"]
})

# Descriptive statistics
display(price_comparison.describe())

# Compare values where both prices are available
valid_prices = price_comparison.dropna(
    subset=["price", "quoted_price_per_night"]
)

print("Listings with both prices:", len(valid_prices))

print(
    "Listings with different prices:",
    (
        valid_prices["price"] !=
        valid_prices["quoted_price_per_night"]
    ).sum()
)

display(valid_prices.head(10))
```
